```python
USER = "utilizator"   # numele dumneavoastră de utilizator GitHub
VARIANTA = "S"        # S la simulare; la colocviu, varianta din README
```

## Valoarea N

Depinde de numele de utilizator și de variantă, deci fiecare student are alt rezultat:

$$N = 1000 + \left(k \cdot \sum_{c \in \text{USER}} \operatorname{cod}(c)\right) \bmod 500,$$

unde $\operatorname{cod}(c)$ este codul Unicode al caracterului $c$, iar $k$ depinde de variantă.


```python
K = {"S": 2, "A": 3, "B": 5, "C": 7, "D": 11, "E": 13, "F": 17}


def calculeaza_n(user, varianta):
    s = sum(ord(c) for c in user.strip().lower())
    return 1000 + (K[varianta.strip().upper()] * s) % 500


N = calculeaza_n(USER, VARIANTA)
print(f"REZULTAT: N={N}")
```

## Un grafic (nu se punctează)

Cum variază N cu varianta, pentru numele dumneavoastră de utilizator.


```python
import matplotlib.pyplot as plt

variante = list(K)
plt.bar(variante, [calculeaza_n(USER, v) for v in variante])
plt.title(f"N pentru {USER}")
plt.xlabel("Varianta")
plt.ylabel("N")
plt.show()
```
